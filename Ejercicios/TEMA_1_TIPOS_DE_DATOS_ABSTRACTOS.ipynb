# Tipos de Datos

Un **Tipo de Dato Abstracto (TDA)** es un modelo de datos que define *qué* operaciones se pueden hacer, sin especificar *cómo*. Es como una caja negra: sabes lo que hace, pero no cómo funciona por dentro.

### Ejemplos de Tipos de Datos Abstractos (TDA):

*   **Lista:** Colección ordenada de elementos (añadir, eliminar, acceder por índice).
*   **Pila (Stack):** LIFO (Last In, First Out) – como una pila de platos.
*   **Cola (Queue):** FIFO (First In, First Out) – como una fila de personas.
*   **Árbol:** Estructura jerárquica (con nodos y ramas).
*   **Grafo:** Nodos conectados por aristas (red de carreteras, amistades).

La **Gestión de Memoria** es cómo una computadora administra su memoria. Se encarga de asignar espacio a los programas cuando lo necesitan y de liberarlo cuando terminan, como organizar juguetes en una caja: cada uno tiene su lugar y cuando no se usa, se guarda. Esto asegura que la computadora funcione eficientemente y que los programas no se 'pisen' entre sí.

### Explicación del Problema (DESAFIO 1) (Referencias en Python):

Cuando haces `copy_data = original_data`, no estás creando una nueva lista con los mismos valores, sino que ambas variables (`original_data` y `copy_data`) están apuntando a **la misma lista** en la memoria. Es decir, `copy_data` es solo otro nombre o una 'referencia' al mismo objeto que `original_data`. Por eso, cualquier cambio que hagas a través de `copy_data` se reflejará también en `original_data` porque son el mismo objeto subyacente.

### Solución:

Para crear una copia **independiente** de una lista, de modo que modificar una no afecte a la otra, necesitas realizar una copia superficial (shallow copy) o una copia profunda (deep copy) dependiendo de si los elementos de la lista son mutables o inmutables. Para este caso simple de números, una copia superficial es suficiente.

````python
# CÓDIGO CORREGIDO
original_data = [10, 20, 30]

# Forma correcta de copiar una lista para que sea independiente:
# Opción 1: Usar slicing
independent_copy = original_data[:]

# Opción 2: Usar el constructor list()
# independent_copy = list(original_data)

# Opción 3: Usar el módulo copy (para copias superficiales y profundas)
# import copy
# independent_copy = copy.copy(original_data) # Para shallow copy
# independent_copy = copy.deepcopy(original_data) # Para deep copy (si hay objetos anidados mutables)

independent_copy.append(40)

print(f"Original (corregida): {original_data}")
print(f"Copia independiente: {independent_copy}")
````

### Explicación Simplificada del Desafío 2 Resuelto: Implementación de TDA Pila (Stack)

El Desafío 2 pedía corregir una clase `Stack` para que funcionara como una pila real (LIFO - Último en entrar, primero en salir).

Los **errores principales** en la clase original eran:

1.  **`is_empty`:** Su lógica estaba invertida; decía que la pila estaba llena cuando estaba vacía, y viceversa.
2.  **`push`:** No agregaba elementos a la pila, sino que reemplazaba todos los elementos existentes por el nuevo.
3.  **`pop`:** Intentaba quitar un elemento sin verificar si la pila tenía alguno, y además no devolvía el elemento que quitaba.

Las **correcciones aplicadas** fueron:

*   **`is_empty`:** Se ajustó para que devolviera `True` cuando la pila *realmente* no tuviera elementos.
*   **`push`:** Ahora añade correctamente el nuevo elemento al final de la lista (`append`), simulando que se coloca en la cima de la pila.
*   **`pop`:** Se mejoró para que primero verifique si la pila tiene elementos. Si los tiene, usa `pop()` para quitar y devolver el último elemento (el de la cima). Si no hay elementos, indica que la pila está vacía para evitar errores.

````python
class Stack:
    def __init__(self):
        self.items = []

    def is_empty(self):
        # CORREGIDO: Retorna True si la pila está vacía
        return len(self.items) == 0

    def push(self, item):
        # CORREGIDO: Agrega el item al final de la lista (cima de la pila)
        self.items.append(item)

    def pop(self):
        # CORREGIDO: Verifica si hay elementos antes de eliminar
        # y retorna el elemento eliminado (el último en entrar)
        if self.is_empty():
            return "La pila está vacía"
        return self.items.pop()

    def peek(self):
        # Opcional: Para ver el elemento superior sin eliminarlo
        if self.is_empty():
            return "La pila está vacía"
        return self.items[-1]

    def size(self):
        return len(self.items)

# --- Prueba corregida ---
mi_pila = Stack()
print("¿Está vacía al inicio?", mi_pila.is_empty()) # Debe ser True

mi_pila.push("A")
print(f"Pila después de push('A'): {mi_pila.items}")

mi_pila.push("B")
print(f"Pila después de push('B'): {mi_pila.items}")

print("¿Está vacía ahora?", mi_pila.is_empty()) # Debe ser False

print("Elemento sacado (debe ser B):", mi_pila.pop())
print(f"Pila después de pop(): {mi_pila.items}")

print("Elemento sacado (debe ser A):", mi_pila.pop())
print(f"Pila después de pop(): {mi_pila.items}")

print("¿Está vacía al final?", mi_pila.is_empty()) # Debe ser True

print("Intentando sacar de una pila vacía:", mi_pila.pop())
````

### Desafío 3 Resuelto: Gestión de Memoria en el Heap (Nodos Dinámicos)

````python
class Nodo:
    def __init__(self, valor):
        self.valor = valor
        self.siguiente = None # Este es nuestro 'puntero' al siguiente espacio en el Heap

# --- Prueba Corregida ---
# El estudiante intenta crear una cadena: [Nodo 1] -> [Nodo 2]

# PASO 1: Crear el primer nodo y guardar su referencia
primer_nodo = Nodo("Datos Importantes 1")
contenedor = primer_nodo # 'contenedor' ahora apunta al primer nodo

# PASO 2: Crear el segundo nodo
segundo_nodo = Nodo("Datos Importantes 2")

# PASO 3: Enlazar el primer nodo al segundo
primer_nodo.siguiente = segundo_nodo

# Verificación
print(f"Contenido del primer nodo (a través de contenedor): {contenedor.valor}")

if contenedor.siguiente is not None:
    print(f"Contenido del segundo nodo (enlazado): {contenedor.siguiente.valor}")
else:
    print("ERROR: El primer nodo no apunta al segundo.")

print("¡Referencia al primer nodo mantenida y nodos enlazados correctamente!")
````

### Explicación Simplificada del Desafío 3 Resuelto: Gestión de Memoria en el Heap (Nodos Dinámicos)

El Desafío 3 trataba sobre cómo conectar correctamente objetos (`Nodo`s) en Python para formar una secuencia, manteniendo siempre un "punto de entrada" a esa secuencia.

El **problema original** era que al crear el segundo nodo, el código sobrescribía la variable que apuntaba al primer nodo. Esto hacía que perdiéramos la conexión con el inicio de nuestra cadena de nodos, dejando el primer nodo inaccesible.

La **solución** consistió en:

1.  Crear el `primer_nodo` y guardar su referencia en una variable (`primer_nodo` y también `contenedor`).
2.  Crear el `segundo_nodo` de forma independiente.
3.  Conectar estos dos nodos haciendo que el `primer_nodo` apunte al `segundo_nodo` a través de su atributo `siguiente` (`primer_nodo.siguiente = segundo_nodo`).

De esta manera, la variable `contenedor` siempre nos permite acceder al `primer_nodo`, y desde el `primer_nodo` podemos seguir la cadena para llegar al `segundo_nodo`.

### Desafío 4 Resuelto: TDA Punto y Circunferencia (Paso por Referencia vs. Valor)

````python
class Punto:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"

class Circunferencia:
    def __init__(self, centro, radio):
        # CORREGIDO: Creamos una nueva instancia de Punto con los valores del centro original.
        # Esto asegura que la circunferencia tenga su propia copia de las coordenadas.
        self.centro = Punto(centro.x, centro.y)
        self.radio = radio

    def __str__(self):
        return f"Circunferencia con centro en {self.centro} y radio {self.radio}"

# --- Prueba Corregida ---
mi_punto_original = Punto(1, 2)
mi_circunferencia = Circunferencia(mi_punto_original, 5)

print(f"Circunferencia inicial: {mi_circunferencia}")
print(f"Punto original (antes de modificar): {mi_punto_original}")

# El alumno modifica el punto original
mi_punto_original.x = 10
mi_punto_original.y = 20

print(f"Punto original (después de modificar): {mi_punto_original}")
print(f"Circunferencia después de modificar el punto original: {mi_circunferencia}")

# Verificación: El centro de la circunferencia debería seguir siendo (1, 2),
# mientras que mi_punto_original ahora es (10, 20).

if str(mi_circunferencia.centro) == "(1, 2)":
    print("¡Éxito! El centro de la circunferencia se mantuvo inalterado.")
else:
    print("ERROR: El centro de la circunferencia cambió incorrectamente.")
````

### Explicación Simplificada del Desafío 4 Resuelto: TDA Punto y Circunferencia (Paso por Referencia vs. Valor)

El Desafío 4 ilustró un concepto importante en Python: cómo se manejan los objetos mutables cuando se pasan a funciones o constructores (paso por referencia).

El **problema original** era que al crear una `Circunferencia` y pasarle un objeto `Punto` como su centro, la `Circunferencia` no guardaba una *copia* de ese `Punto`, sino una *referencia* al `Punto` original. Esto significa que si el `Punto` original se modificaba (cambiando sus coordenadas `x` o `y`), el centro de la `Circunferencia` también cambiaba inesperadamente, porque ambos estaban apuntando al mismo objeto `Punto` en la memoria.

La **solución** se implementó en el método `__init__` de la clase `Circunferencia`.

En lugar de simplemente hacer `self.centro = centro`, que copiaría la referencia, se modificó a:

```python
self.centro = Punto(centro.x, centro.y)
```

Con esta corrección, cuando se inicializa una `Circunferencia`, se crea un **nuevo** objeto `Punto` utilizando las coordenadas (`x` e `y`) del `Punto` que se le pasó. De esta manera, la `Circunferencia` tiene su *propia* instancia de `Punto` para su centro, independiente del objeto `mi_punto_original`. Así, cualquier cambio posterior a `mi_punto_original` no afectará al centro de la `Circunferencia`, resolviendo el problema del paso por referencia.

### Desafío 5: TDA Automóvil y Garaje (Gestión de Objetos en Colecciones)

**Contexto:** Estamos modelando un `Automovil` y un `Garaje` que almacena una colección de automóviles. Cuando se trabaja con listas de objetos, es fácil cometer errores al añadir o modificar elementos, especialmente si no se comprende cómo Python maneja las referencias a objetos en el *heap*.

**Problema:** Un estudiante intenta añadir varios automóviles a un garaje. Sin embargo, debido a una reutilización incorrecta de una variable de objeto `Automovil` al añadir a la lista, termina con un garaje que parece contener el mismo automóvil repetido varias veces, en lugar de automóviles distintos.

````python
class Automovil:
    def __init__(self, marca, modelo, color):
        self.marca = marca
        self.modelo = modelo
        self.color = color

    def __str__(self):
        return f"{self.color} {self.marca} {self.modelo}"

class Garaje:
    def __init__(self):
        self.automoviles = []

    def agregar_automovil(self, auto):
        self.automoviles.append(auto)

    def mostrar_automoviles(self):
        print("Automóviles en el garaje:")
        for i, auto in enumerate(self.automoviles):
            print(f"  {i+1}. {auto}")

# --- Prueba Corregida ---
mi_garaje = Garaje()

# CORREGIDO: Crear una nueva instancia de Automovil para cada coche distinto

# Primer automóvil
auto1 = Automovil("Toyota", "Corolla", "Rojo")
mi_garaje.agregar_automovil(auto1)

# Segundo automóvil (¡una nueva instancia!)
auto2 = Automovil("Honda", "Civic", "Azul")
mi_garaje.agregar_automovil(auto2)

# Tercer automóvil (otra nueva instancia)
auto3 = Automovil("Ford", "Focus", "Verde")
mi_garaje.agregar_automovil(auto3)

mi_garaje.mostrar_automoviles()

# Verificación: Cada automóvil en el garaje debe ser una instancia única
if len(mi_garaje.automoviles) == 3 and mi_garaje.automoviles[0] is not mi_garaje.automoviles[1] and mi_garaje.automoviles[0].marca == "Toyota" and mi_garaje.automoviles[1].marca == "Honda" and mi_garaje.automoviles[2].marca == "Ford":
    print("¡Éxito! El garaje contiene automóviles distintos con sus propiedades correctas.")
else:
    print("ERROR: Los automóviles en el garaje no son distintos o sus propiedades son incorrectas.")
````

### Desafío 6: TDA Línea (Inmutabilidad Conceptual con Objetos Mutables)

**Contexto:** Una `Línea` se define por dos `Punto`s: un punto de inicio y un punto final. Conceptualmente, una vez que una línea se define, sus puntos extremos no deberían cambiar si se modifican los `Punto`s originales que se usaron para crearla, a menos que se quiera explícitamente.

**Problema:** Similar al desafío de la circunferencia, si la clase `Línea` almacena referencias directas a los objetos `Punto` mutables pasados durante su inicialización, cualquier modificación externa a esos `Punto`s afectará la `Línea` de forma inesperada. El estudiante espera que la línea permanezca estática después de su creación.

````python
class Punto:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"

class Linea:
    def __init__(self, inicio, fin):
        # CORREGIDO: Creamos nuevas instancias de Punto con los valores de los puntos originales.
        # Esto asegura que la línea tenga sus propias copias de las coordenadas de inicio y fin.
        self.inicio = Punto(inicio.x, inicio.y)
        self.fin = Punto(fin.x, fin.y)

    def __str__(self):
        return f"Línea de {self.inicio} a {self.fin}"

# --- Prueba Corregida ---
punto_a = Punto(0, 0)
punto_b = Punto(5, 5)

mi_linea = Linea(punto_a, punto_b)
print(f"Línea original: {mi_linea}")

# El alumno modifica uno de los puntos originales
punto_a.x = 10
punto_a.y = 10

print(f"Línea después de modificar el punto 'A': {mi_linea}")

# Verificación: Los puntos de la línea deberían seguir siendo (0, 0) y (5, 5),
# mientras que 'punto_a' ahora es (10, 10).

if str(mi_linea.inicio) == "(0, 0)" and str(mi_linea.fin) == "(5, 5)":
    print("¡Éxito! Los puntos de la línea se mantuvieron inalterados.")
else:
    print("ERROR: Los puntos de la línea cambiaron incorrectamente.")
````
